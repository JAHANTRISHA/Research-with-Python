```python
# Your tasks:
# 1. Calculate mean, median, std, min, max for each subject using NumPy
# 2. Find total students who scored above 80 in math
# 3. Find total students who scored below 50 in any subject
# 4. Calculate the overall average score across all three subjects
# Expected output example:
# Math - Mean: 66.1, Median: 66.0, Std: 15.2
# Students with math > 80: 132
# Students with any score < 50: 178
```

```python
#Task 1.2: NumPy Array Operations
# Convert score columns to numpy arrays
math_scores = df['math score'].values
reading_scores = df['reading score'].values
writing_scores = df['writing score'].values


# 1. Calculate mean, median, std, min, max for each subject using NumPy
print(f"Math - Mean: {round(np.mean(math_scores), 1)}, Median: {round(np.median(math_scores), 1)}, Std: {round(np.std(math_scores), 1)}, Min: {np.min(math_scores)}, Max: {np.max(math_scores)}")
print(f"Reading - Mean: {round(np.mean(reading_scores), 1)}, Median: {round(np.median(reading_scores), 1)}, Std: {round(np.std(reading_scores), 1)}, Min: {np.min(reading_scores)}, Max: {np.max(reading_scores)}")
print(f"Writing - Mean: {round(np.mean(writing_scores), 1)}, Median: {round(np.median(writing_scores), 1)}, Std: {round(np.std(writing_scores), 1)}, Min: {np.min(writing_scores)}, Max: {np.max(writing_scores)}")
print("")

# 2. Find total students who scored above 80 in math
high_math_score = math_scores > 80
print("Total students who scored above 80 in math:", np.sum(high_math_score))
print("")

# 3. Find total students who scored below 50 in any subject
lowest_scores = (math_scores < 50) | (reading_scores < 50) | (writing_scores < 50)
print("Total students who scored below 50 in any subject: ", np.sum(lowest_scores))
print("")

# 4. Calculate the overall average score across all three subjects
average_scores = df[['math score', 'reading score', 'writing score']]
all_three_average_score = average_scores.mean()
print("the overall average score across all three subjects: ", round(all_three_average_score.mean()))
```

```python
#Task 2.2: Create and Fix Missing Value
# Artificially create missing data for practice
import numpy as np

# Your tasks:
# 1. Make a copy of the dataset: df_practice = df.copy()
# 2. Randomly set 50 math scores to np.nan
# 3. Randomly set 30 writing scores to np.nan
# 4. Fill missing math scores with the MEDIAN math score
# 5. Fill missing writing scores with the MEAN writing score
# 6. Verify no missing values remain
```

```python
# Steps to follow:
# 1. Make a copy of the dataset: df_practice = df.copy()
# Step 1: Create copy and show original missing values
df_practice = df.copy()
print("Original missing values:", df_practice.isnull().sum())
print("***********************************")
# Step 2: Create missing data
np.random.seed(42)  # For reproducible results

# 2. Randomly set 50 math scores to np.nan
missing_math_idx = np.random.choice(df_practice.index, 50, replace=False)
df_practice.loc[missing_math_idx, "math score"] = np.nan
print("\nTotal missing math scores:", df_practice['math score'].isna().sum())
print("---------------------------------------------------")

# 3. Randomly set 30 writing scores to np.nan
missing_writing_idx = np.random.choice(df_practice.index, 30, replace=False)
df_practice.loc[missing_writing_idx, "writing score"] = np.nan
print("\nTotal missing writing scores:", df_practice['writing score'].isna().sum())
print("---------------------------------------------------")

# 4. Fill missing math scores with the MEDIAN math score
math_median = df_practice["math score"].median()
df_practice["math score"] = df_practice["math score"].fillna(math_median)
print(f"\nFilled missing math scores with median ({math_median}): \n", df_practice.isnull().sum())
print("---------------------------------------------------")

writing_mean = df_practice["writing score"].mean()
df_practice["writing score"] = df_practice["writing score"].fillna(writing_mean)
print(f"\nFilled missing writing scores with mean ({writing_mean}):\n", df_practice.isnull().sum())
print("---------------------------------------------------")

print("\nFinal check, no missing values remain:\n", df_practice.isnull().sum())
```

```python
#Task 3.2: Basic Groupby Analysis
# Comprehensive groupby analysis
# Your tasks:

# 1. Group by 'gender' and calculate:
#    - Average scores in all subjects
#    - Count of students

# 2. Group by 'test preparation course' and show:
#    - Mean scores for each subject
#    - Show which subject benefits most from test prep

# 3. Group by 'parental level of education' and calculate:
#    - Average math score for each education level
#    - Rank education levels by math performance

# 4. Group by 'lunch' type and show:
#    - Average scores for each subject
#    - Count of students in each lunch category
```

```python
# 1. Group by 'gender' and calculate:
#    - Average scores in all subjects
#    - Count of students
Average_scores = df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()
print(f"Average scores in all subjects: \n {Average_scores}")
students = df.groupby('gender')[['gender']].count()
print(f"\nCount of students: \n {students}")
print("-------------------------------------------------------------")

# 2. Group by 'test preparation course' and show:
#    - Mean scores for each subject
#    - Show which subject benefits most from test prep
Mean_scores = df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()
print(f"Mean scores for each subject: \n {Mean_scores}")
mean_scores_difference = Mean_scores.loc['completed'] - Mean_scores.loc['none']
most_benefited_subject = mean_scores_difference.idxmax()
print(f"\n Subject benefits most from test prep: \n {most_benefited_subject}")
print("***************************************************************")

# 3. Group by 'parental level of education' and calculate:
#    - Average math score for each education level
#    - Rank education levels by math performance
Average_Math_scores = df.groupby('parental level of education')[['math score']].mean()
print(f"Average math score for each education level: \n {Average_Math_scores}")
ranked_levels = Average_Math_scores.sort_values(by='math score', ascending=False)
print(f"\nRank education levels by math performance: \n {ranked_levels}")
print("-----------------------------------------------------------------")

# 4. Group by 'lunch' type and show:
#    - Average scores for each subject
#    - Count of students in each lunch category
Average_scores_lunch = df.groupby('lunch')[['math score', 'reading score', 'writing score']].mean()
print(f"Average scores for each subject: \n {Average_scores_lunch}")
lunch_category = df.groupby('lunch')[['lunch']].count()
print(f"\nCount of students in each lunch category: \n {lunch_category}")
print("")
```

```python
#Task 3.3: Use your data analysis skills to answer these easy questions:

# 1. Who performs better in math - males or females?
#    - Calculate average math score by gender
#    - Show the difference between male and female average
Math_scores = df.groupby('gender')[['math score']].mean()
print(f"Average math score by gender: \n {Math_scores}")

difference_math_score = Math_scores.loc['male'] - Math_scores.loc['female']
print(f"\n The difference between male and female average: {difference_math_score}")
print("-------------------------------------------------------\n")

# 2. Which subject do students perform best in overall?
#    - Calculate the overall average for math, reading, and writing
#    - Rank the subjects from highest to lowest average score

subjects = df[['math score', 'reading score', 'writing score']]
overall_avg = subjects.mean()
print(f"Calculate the overall average for math, reading, and writing: \n{overall_avg}")
ranked_subjects = overall_avg.sort_values(ascending=False)
print(f"Rank the subjects from highest to lowest average score: \n{ranked_subjects}")
print("*****************************************************\n")

# 3. What's the impact of parental education?
#    - Find the average total score for each parental education level
#    - Identify which education level leads to highest student performance
df['total_score'] = df[['math score', 'reading score', 'writing score']].sum(axis=1)
parental_education_level = df.groupby('parental level of education')['total_score'].mean()
parental_education_level_sorted = parental_education_level.sort_values(ascending=False)
print(f"the average total score for each parental education level: \n {parental_education_level_sorted}")
print(f"\nIdentify which education level leads to highest student performance: \n", parental_education_level.idxmax())
print("")
```
