```python
#Load the Data
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# TODO: Load the housing data
df = pd.read_csv('/content/drive/MyDrive/skillMorph/Class5/Housing.csv')

# TODO: How many houses are there?
print(f"Number of houses: {len(df['price'])}")  # Fill this

# TODO: Show first 3 houses
print("First 3 houses of the dataset:")
print(df.head(3))
```

```python
# Look at Columns
# TODO: Show all column names
print("Columns:")
print(df.columns.tolist())  # Column names
print("----------------------------------------------\n")

# TODO: Show data types of columns
print("Data types:")
print(df.dtypes)
```

```python
#Convert Yes/No to 1/0
# Check what's in our data
print("Current mainroad values:", df['mainroad'].head())
# Shows: yes, no, yes, yes, no
print("-------------------------------------------------------")
# Models need numbers, not text!
# We need: yes → 1, no → 0
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
df['mainroad_encoded'] = label_encoder.fit_transform(df['mainroad'])
print("\nAfter Label Encoding 'mainroad' or Convert Yes/No to 1/0:")
print(df[['mainroad', 'mainroad_encoded']])
```

```python
label_column_list = ['mainroad', 'guestroom', 'basement',
              'hotwaterheating', 'airconditioning', 'prefarea']
encoders = {}
for col in label_column_list:
    label_encoder = LabelEncoder()
    df[col + '_encoded'] = label_encoder.fit_transform(df[col])
    print(f"\nAfter Label Encoding '{col}':")
    print(df[[col, col + '_encoded']].head(3))
    encoders[col] = label_encoder
```

```python
import pandas as pd

# One-hot encode furnishingstatus
furnishingstatus_encoded = pd.get_dummies(df['furnishingstatus'], prefix='furnishingstatus')

# Save categories for later use
furnishingstatus_categories = furnishingstatus_encoded.columns.tolist()

# Concatenate to the main dataframe
df = pd.concat([df, furnishingstatus_encoded], axis=1)

print("\nOne-Hot Encoding for 'furnishingstatus':")
print(furnishingstatus_encoded.head())

print("\nFinal data with all encodings:")
print(df.head(5))

print("\nCategories remembered for later use:")
print(furnishingstatus_categories)
```

```python
# Split into Train and Test
# TODO: Split - 80% train, 20% test

X = df.drop(['price', 'mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea', 'furnishingstatus'], axis=1)
y = df['price']                # Only price

print(X.columns.to_list)
print(f"X has all houses information: \n{X.head(2)}")
print(f"y has price of houses: {y.head(2)}")

# Split: 80% for training, 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,    # 20% for test
    random_state=20   # Same result every time
)

print(f"\nWe will train with: {len(X_train)} houses information")
print(f"We will test with: {len(X_test)} houses information")
```

```python
#Apply StandardScaler
# Area is huge (1650-13300), Bedrooms is small (1-6)!
scaler = StandardScaler()

# Learn from training data and scale it
X_train_scaled = scaler.fit_transform(X_train)

# Scale test data (just transform, don't fit)
X_test_scaled = scaler.transform(X_test)

print("\nAfter scaling:")
print("All features now have similar range!\n")
print(f"Example - First house's price before scaling: {X_train.values[0][0]}")
print(f"Example - First house's price after scaling: {X_train_scaled[0][0]:.2f}")
```

```python
# price is continuous numeric and we do't have to predict yes/no. It's a regression problem. Therefore, LinearRegression model is used.
#train and test model
from sklearn.linear_model import LinearRegression

# Create model
model = LinearRegression()

# Train the model
model.fit(X_train_scaled, y_train)
print("✅ Model trained!")

# Check accuracy on training data
train_score = model.score(X_train_scaled, y_train)
print(f"Training accuracy: {train_score:.2%}")

# Check accuracy on test data
test_score = model.score(X_test_scaled, y_test)
print(f"Test accuracy: {test_score:.2%}")
```

```python
from sklearn.model_selection import cross_val_score

# 10-fold cross-validation
scores = cross_val_score(
    model,           # Our model
    X_train_scaled,  # Training data
    y_train,         # Training labels
    cv=10            # 10 mini-tests
)

print("10 mini-test scores:")
for i in range(10):
    print(f"  Test {i+1}: {scores[i]:.2%}")

print(f"\nAverage: {scores.mean():.2%}")
print(f"This means our model is {scores.mean():.2%} accurate!")
```
