```python
# 2.10: Write a program that checks eligibility for a discount (student AND under 25, OR senior over 60).
age = int(input("Enter your age: "))
status = input("Are you a student or senior: ")
if status == "student" and age < 25:
    print("You are eligibility for a discount")
elif status == "senior" and age < 60:
    print("You are not eligibility for a discount")
elif status == "senior" and age > 60:
    print("You are eligibility for a discount")
else:
    print("You are not eligibility for a discount")
```

```python
# 2.11: Write a program that prints the first 10 even numbers using a for or while loop.
count = 0

print("Even numbers:")
for i in range(1, 50):
    if count<10:
        if i % 2 == 0 :
            print(i, end=" ")
            count=count+1
total=count
print(f"\nFirst {total} even numbers.")
```

```python
# 2.12: Write a program that prints a simple number pyramid pattern using nested loops.

for i in range(1, 9):
    for j in range(9-i):
        print(" ", end="")
    for k in range(2 * i - 1):
        print("7", end="")
    print("")
```

```python
# 2.13: Create a function that converts temperature from Celsius to Fahrenheit using the formula: F = (C × 9/5) + 32

def temperature(C):
    F = (C * 9/5) + 32
    return F
result = temperature(28)
print(f"After converting the temperature in Fahrenheit, the result is: {result}")
```
